```python
#cleaned up from above

import os
import subprocess
from pathlib import Path

def process_pdf(pdf_path, output_dir=None, language="ch", depth=1, model="auto"):
    """
    Process a PDF file through magic-pdf with the specified parameters.
    
    Parameters:
    - pdf_path: Path to the PDF file
    - output_dir: Directory to save output (defaults to ~/mineru/ if None)
    - language: Language code (default: 'ch' for Chinese & English)
    - depth: Processing depth (default: 1)
    - model: Model to use (default: 'auto')
    """
    # Normalize paths
    pdf_path = os.path.expanduser(pdf_path)
    
    if output_dir is None:
        output_dir = os.path.expanduser("~/mineru/")
    else:
        output_dir = os.path.expanduser(output_dir)
    
    # Ensure output directory exists
    Path(output_dir).mkdir(parents=True, exist_ok=True)
    
    # Build and execute the command
    cmd = f'magic-pdf -p "{pdf_path}" -o "{output_dir}" -m {model} -l {language} -d {depth}'
    print(f"Executing: {cmd}")
    
    return subprocess.run(cmd, shell=True)

# Example usage:
# Process a single PDF
process_pdf("/Users/lukasfiller/Library/CloudStorage/OneDrive-DKIAsia-PacificCenterforSecurityStudies/APCSS_Docs/topics/UVa - PRC Technology Security/implement challenges/logisitics_policy_summary_2024_zh.pdf")

# Process multiple PDFs
#pdfs_to_process = [
#    "~/path/to/first/document.pdf",
#    "~/path/to/second/document.pdf"
#]

#for pdf in pdfs_to_process:
#    process_pdf(pdf)
```

```python
from markdownify import markdownify as md

# Function to convert HTML content to Markdown and save
def html_to_markdown(data, output_file="content.md"):
    # Extract the HTML content
    html_content = data.get("html", "")
    title = data.get("title", "Untitled")
    base_url = data.get("base_url", "Unknown Source")
    
    # Convert HTML to Markdown
    markdown_content = md(html_content)
    
    # Combine title, source, and content into Markdown format
    markdown = f"# {title}\n\n"
    markdown += f"**Source:** [Original Article]({base_url})\n\n"
    markdown += markdown_content
    
    # Write to a Markdown file
    with open(output_file, "w", encoding="utf-8") as file:
        file.write(markdown)
    print(f"Markdown content saved to {output_file}")

# Example extracted data from magic-html
data = {
    "title": "From Cognition to Action: Open Source Intelligence Development",
    "html": "<p>This is an example article discussing open source intelligence...</p>",
    "base_url": "https://example.com"
}

# Convert HTML to Markdown and save
html_to_markdown(data)
```
